```python
%matplotlib widget
# you may need to 'pip install ipympl' to get the animation to work
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
```

### Advection

```python
n = 100

f = np.zeros(n)
x = np.linspace(0,1,n)

i1 = int(0.1*n)
i2 = int(0.2*n)
f[i1:i2] = 1.0
fp = f.copy()

alpha = 0.5

plt.close()
fig, ax = plt.subplots()
dat, = ax.plot(x, f)
ax.set_ylim(-0.5, 1.5)

def update(k):
    global f

    # FTCS update
    # we assume periodic boundary conditions, so use numpy.roll to get adjacent grid points
    fnew = f - 0.5 * alpha * (np.roll(f,-1) - np.roll(f,1))
    f = fnew

    dat.set_ydata(f)
    print('Step %d' % (k,))
    ax.set_title('Step %d' %(k,))
    return dat,

ani = FuncAnimation(fig, update, frames=100, interval=10, repeat=False)
plt.show()
```

### Diffusion

```python
n = 100
f = np.zeros(n)
x = np.linspace(0,1,n)

# initial condition
f[int(0.4*n):int(0.6*n)] = 1.0

alpha = 0.5

plt.close()
fig, ax = plt.subplots()
dat, = ax.plot(x, f)
ax.set_ylim(-0.5, 1.5)

explicit = True

A_exp = np.eye(n)*(1.0-2.0*alpha) + np.eye(n,k=1)*alpha + np.eye(n,k=-1)*alpha
A_imp = np.eye(n)*(1.0+2.0*alpha) + np.eye(n,k=1)*(-alpha) + np.eye(n,k=-1)*(-alpha)

def update(k):
    global f,A_exp,A_imp

    if explicit:
        # explicit
        fnew = A_exp@f
        # or equivalently
        #fnew = f + alpha * (np.roll(f,-1) - 2*f + np.roll(f,1))
    else:
        # implicit
        fnew = np.linalg.solve(A_imp,f)

    f = fnew
    
    dat.set_ydata(f)
    print('Step %d' % (k,))
    ax.set_title('Step %d' %(k,))
    return dat,

ani = FuncAnimation(fig, update, frames=100, interval=10, repeat=False)
plt.show()
```
